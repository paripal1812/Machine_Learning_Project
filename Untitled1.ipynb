```python
df['chol'].value_counts()
```

```python
ch_mean = df.loc[df['chol'] != 0, 'chol'].mean()
```

```python
df['chol'] = df['chol'].replace(0,ch_mean)
df['chol'] = df['chol'].round(2)
```

```python
ch_mean = df.loc[df['trestbps'] != 0, 'trestbps'].mean()
df['trestbps'] = df['trestbps'].replace(0,ch_mean)
df['trestbps'] = df['trestbps'].round(2)
```

```python
def plotting(var,num):
    plt.subplot(2,2,num)
    sns.histplot(df[var],kde = True)

plotting('age',1)
plotting('oldpeak',2)
plotting('chol',3)
plotting('trestbps',4)

plt.tight_layout()
```
